```python
#输入月份数字，输出月份英文缩写
months="JanFebMarAprMayJunJulAugSepOctNovDec"
n=input("请输入月份的数字(1-12):")
#子字符串的起始位置
pos=(int(n)-1) * 3
monthAbbrev=months[pos:pos+3]
print("第"+n+"月份的缩写为："+monthAbbrev+".")
```

```python
#pi
from random import random
from math import sqrt
#DARTS 总点数，hits圆内点数量
DARTS = 2**20
hits = 0

for i in range(1,DARTS):
    #生成随机点
    x, y = random(), random()
    #计算点到圆心的距离
    dist = sqrt(x**2 + y**2)
    if dist <= 1.0:
        #圆内点hits+1
        hits = hits + 1
pi = 4 * (hits/DARTS)
print("Pi is %s" % pi)
```

```python
#求cos函数面积
#蒙特卡洛求sin
from random import random
from math import cos
import numpy as np


DARTS=2**20
hits=0

for i in range(1,DARTS):
    x,y=0.5*(np.pi)*random(),random()
    if(y<=cos(x)):
        hits=hits+1
        
s=np.pi*hits/DARTS
print(s)
```

```python
import math
print("一元二次方程求解")
a,b,c = eval(input("输入系数a,b,c:"))
delta= b*b-4*a*c
if delta >= 0:
    delta2= math.sqrt(delta)
    root1= (-1*b+delta2)/(2*a)
    root2= (-1*b-delta2)/(2*a)
    print ("方程解为：",root1,root2)
else:
    print("方程无实根")
```

```python
#简单置换加密 k=3
#获得需要加密的M
plaintext = input("请输入加密消息:")
#建立映射表 
plaint_list ="abcdefghijklmnopqrstuvwxyz"
cipher_list = "DEFGHIJKLMNOPQRSTUVWXYZABC"
#输入信息
ciphertext =""
if len(plaintext)>0:
    #对消息中的每个字母进行查询
    for x in plaintext:
        #获得位置索引
        index = plaint_list.find(x)
        #通过索引获得替换字母
        ciphertext =ciphertext+cipher_list[index]
        #输出密文结果    
    if len(ciphertext) > 0:
        print("明文序列为:"+plaint_list)
        print("密文序列为:"+cipher_list)
        print("消息为:"+plaintext)
        print("密文为:"+ciphertext)
            
else:
    print("输入错误")
```

```python
# 编程实现解密算法
#简单置换加密 k=3
#获得需要解密的密文
ciphertext = input("请输入需解密的消息:")
#建立映射表 
plaint_list ="abcdefghijklmnopqrstuvwxyz"
cipher_list = "DEFGHIJKLMNOPQRSTUVWXYZABC"
#输入信息
plaintext =""
if len(ciphertext)>0:
    #对消息中的每个字母进行查询
    for x in ciphertext:
        #获得位置索引
        index = cipher_list.find(x)
        #通过索引获得替换字母
        plaintext =plaintext+plaint_list[index]
        # plaintext += plaint_list[index]
        #输出密文结果    
    if len(plaintext) > 0:
        print("明文序列为:"+plaint_list)
        print("密文序列为:"+cipher_list)
        print("解密密文为:"+ciphertext)
        print("解密结果为:"+plaintext)
            
else:
    print("输入错误")
```
