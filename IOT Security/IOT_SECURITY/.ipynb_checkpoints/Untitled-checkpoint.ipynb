```python
#pi
from random import random
from math import sqrt
#DARTS 总点数，hits圆内点数量
DARTS = 2**20
hits = 0

for i in range(1,DARTS):
    #生成随机点
    x, y = random(), random()
    #计算点到圆心的距离
    dist = sqrt(x**2 + y**2)
    if dist <= 1.0:
        #圆内点hits+1
        hits = hits + 1
pi = 4 * (hits/DARTS)
print("Pi is %s" % pi)
```

```python
#求cos函数面积
#蒙特卡洛求sin
from random import random
from math import cos
import numpy as np


DARTS=2**20
hits=0

for i in range(1,DARTS):
    x,y=0.5*(np.pi)*random(),random()
    if(y<=cos(x)):
        hits=hits+1
        
s=np.pi*hits/DARTS
print(s)
```

```python
import math
print("一元二次方程求解")
a,b,c = eval(input("输入系数a,b,c:"))
delta= b*b-4*a*c
if delta >= 0:
    delta2= math.sqrt(delta)
    root1= (-1*b+delta2)/(2*a)
    root2= (-1*b-delta2)/(2*a)
    print ("方程解为：",root1,root2)
else:
    print("方程无实根")
```
