```python
# 223428010210 陈梓欣

# 维吉尼亚加密算法 Vigenère Cipher
# 检查输入是否为全字母
def input_chk(message):
    chk_list = 'ABCDEFGHIJKLMNOPQRSTUVWXYZ'
    for x in message.upper():
        if x not in chk_list:
            return False
    return True
# 凯撒加密（单字符）
def caesar_encrypt_char(key, ch):
    plaint_list = "abcdefghijklmnopqrstuvwxyz"
    cipher_list = "ABCDEFGHIJKLMNOPQRSTUVWXYZABCDEFGHIJKLMNOPQRSTUVWXYZ"
    index = plaint_list.find(ch)
    if index == -1:
        return ch  # 非字母直接返回
    return cipher_list[key + index]
# 维吉尼亚加密
def vigenere_encrypt(key, message):
    strlist = "abcdefghijklmnopqrstuvwxyz"
    key = key.lower()
    message = message.lower()
    cipher = ""

    key_len = len(key)
    for i in range(len(message)):
        k = key[i % key_len]  # 循环使用密钥
        key_index = strlist.find(k)
        m_char = message[i]
        cipher_char = caesar_encrypt_char(key_index, m_char)
        cipher += cipher_char
    return cipher
# 主函数
def main():
    key = input("请输入加密密钥（字母）: ")
    message = input("请输入加密消息（字母）: ")

    if not (input_chk(key) and input_chk(message)):
        print("输入错误：密钥和消息必须只包含英文字母")
        return

    cipher = vigenere_encrypt(key, message)
    print("\n====================")
    print("明文消息: " + message)
    print("密钥:     " + key)
    print("加密结果: " + cipher)
    print("====================")

if __name__ == "__main__":
    main()

       
```

```python
# 223428010210 陈梓欣

# 仿射解密算法实现
# 解密公式：D(y) = a_inv * (y - k) mod 26
def mod_inverse(a, m):
    """
    求 a 关于 m 的模逆元，即 a * a_inv ≡ 1 mod m
    使用扩展欧几里得算法
    """
    for x in range(1, m):
        if (a * x) % m == 1:
            return x
    return None  # 如果没有逆元，返回 None
# 输入密钥 a 和 k
a, k = eval(input("请输入解密密钥 a, k（1~25）："))
# 求 a 的模 26 逆元
a_inv = mod_inverse(a, 26)
if a_inv is None:
    print("错误：密钥 a 与 26 不互质，无法解密！")
    exit()
# 输入密文
ciphertext = input("请输入要解密的密文（大写英文字母）: ")
# 字母表
cipher_list = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
plain_list = "abcdefghijklmnopqrstuvwxyz"
# 解密结果
plaintext = ""
# 遍历密文字符
for ch in ciphertext:
    if ch in cipher_list:
        index = cipher_list.find(ch)              # 找到密文字母索引（0~25）
        pos = (a_inv * (index - k)) % 26          # 解密计算
        plaintext += plain_list[pos]              # 对应小写明文字母
    else:
        plaintext += ch                           # 非字母直接加入
# 输出明文
print("密文为:", ciphertext)
print("解密后的原文为:", plaintext)
```

```python
#223428010210 陈梓欣
# 223428010103 刘渤

#Blundo 传感器密钥分配方案
#服务器端
from random import randint
# 随机生成 199~999 范围内的一个素数作为模数 p
def randprime():
    top = randint(199, 999)
    for x in range(top, 200, -1):
        for y in range(2, x):
            if x % y == 0:
                break
        else:
            return x
# 中心节点：服务器端
class Server:
    def __init__(self):
        self.a = randint(10, 99)
        self.b = randint(10, 99)
        self.c = randint(10, 99)
        self.p = randprime()
        self.chd_index = 0  # 子节点编号计数器
        self.IDs = {}       # 记录所有子节点的 ID

    def __str__(self):
        return f"对称多项式为：f(x,y) = {self.a} + {self.b}*(x+y) + {self.c}*x*y mod {self.p}"
    # 生成多项式的描述
    def poly_generate(self):
        return f"生成多项式为：{self.a} + {self.b}*(x+y) + {self.c}*x*y mod {self.p}"
    # 为一个新子节点生成 ID 和一次函数 A + Bx mod p
    def get_ID(self):
        self.chd_index += 1
        chd_name = 'c' + str(self.chd_index)
        ID = randint(10, 199)
        self.IDs[chd_name] = ID
        A = (self.a + self.b * ID) % self.p
        B = (self.b + self.c * ID) % self.p
        return [chd_name, ID, A, B, self.p]

    def get_p(self):
        return self.p
# 子节点类：客户端
class Client:
    def __init__(self, name, id, A, B, p):
        self.name = name
        self.id = id
        self.A = A
        self.B = B
        self.p = p
        self.recvIDs = {}     # 其他节点 ID 表
        self.shareKeys = {}   # 与其他节点的共享密钥表

    def getID(self):
        return self.id

    def getName(self):
        return self.name

    def __str__(self):
        return f"{self.name}: ID={self.id}, A={self.A}, B={self.B}, p={self.p}"
# 交换两个节点的 ID
def swapKey(c1, c2):
    c1.recvIDs[c2.getName()] = c2.getID()
    c2.recvIDs[c1.getName()] = c1.getID()
# 使用共享一次函数和对方 ID 生成共享密钥
def shareKey(c1, c2):
    id2 = c1.recvIDs[c2.getName()]
    id1 = c2.recvIDs[c1.getName()]
    key1 = (c1.A + c1.B * id2) % c1.p
    key2 = (c2.A + c2.B * id1) % c2.p
    assert key1 == key2, f"密钥不一致！{c1.getName()}={key1}, {c2.getName()}={key2}"
    c1.shareKeys[c2.getName()] = key1
    c2.shareKeys[c1.getName()] = key2
# 方案测试部分
if __name__ == "__main__":
    # 初始化中心服务器
    s = Server()
    print(s)
    print(s.poly_generate())
    # 创建3个子节点
    [name1, id1, A1, B1, p] = s.get_ID()
    c1 = Client(name1, id1, A1, B1, p)
    print(c1)

    [name2, id2, A2, B2, p] = s.get_ID()
    c2 = Client(name2, id2, A2, B2, p)
    print(c2)

    [name3, id3, A3, B3, p] = s.get_ID()
    c3 = Client(name3, id3, A3, B3, p)
    print(c3)
    # 节点交换 ID
    swapKey(c1, c2)
    swapKey(c1, c3)
    swapKey(c2, c3)
    # 节点生成共享密钥
    shareKey(c1, c2)
    shareKey(c1, c3)
    shareKey(c2, c3)
    # 输出共享密钥
    print("\nc1 与其他节点共享密钥：", c1.shareKeys)
    print("c2 与其他节点共享密钥：", c2.shareKeys)
    print("c3 与其他节点共享密钥：", c3.shareKeys)
```

```python
# 223428010210 陈梓欣
# 223428010103 刘渤

# 有限域GF(2^8)乘法定义
def xtime(a):
    return ((a << 1) ^ 0x1B) & 0xFF if a & 0x80 else (a << 1)

def gf_mul(a, b):
    result = 0
    for i in range(8):
        if b & 1:
            result ^= a
        a = xtime(a)
        b >>= 1
    return result

# 列混合（对一列进行）
def mix_single_column(col):
    t = col[0] ^ col[1] ^ col[2] ^ col[3]
    u = col[0]
    col[0] ^= t ^ gf_mul(col[0] ^ col[1], 0x02)
    col[1] ^= t ^ gf_mul(col[1] ^ col[2], 0x02)
    col[2] ^= t ^ gf_mul(col[2] ^ col[3], 0x02)
    col[3] ^= t ^ gf_mul(col[3] ^ u, 0x02)
    return col

# 对整个状态（4x4矩阵）做列混合
def mix_columns(state):
    for i in range(4):
        column = [state[row][i] for row in range(4)]
        mixed = mix_single_column(column)
        for row in range(4):
            state[row][i] = mixed[row]
    return state

# 示例状态矩阵（16字节明文）
state = [
    [0x87, 0xF2, 0x4D, 0x97],
    [0x6E, 0x4C, 0x90, 0xEC],
    [0x46, 0xE7, 0x4A, 0xC3],
    [0xA6, 0x8C, 0xD8, 0x95]
]

print("原始状态矩阵：")
for row in state:
    print([hex(x) for x in row])

mixed_state = mix_columns(state)

print("\n列混合后的状态矩阵：")
for row in mixed_state:
    print([hex(x) for x in row])
```

```python
# 223428010210 陈梓欣
# 223428010103 刘渤

#数字签名
from Crypto import Random
from Crypto.Hash import SHA
from Crypto.Cipher import PKCS1_v1_5 as Cipher_pkcs1_v1_5
from Crypto.Signature import PKCS1_v1_5 as Signature_pkcs1_v1_5

from Crypto.PublicKey import RSA
import base64

# 加密解密：公钥加密，私钥解密
# 签名验签：私钥签名，公钥验签
print ("1、生成私钥和公钥，运行在服务器上，TTP")
# 伪随机数生成器
random_generator = Random.new().read
# rsa算法生成实例
rsa = RSA.generate(1024, random_generator)
# A的私钥对的生成并保存
private_pem = rsa.exportKey()
with open('aaa-private.pem', 'wb') as f:
    f.write(private_pem) 
# A的公钥对的生成并保存
public_pem = rsa.publickey().exportKey()
with open('aaa-public.pem', 'wb') as f:
    f.write(public_pem)
# B的秘钥对的生成并保存
private_pem = rsa.exportKey()
with open('bbb-private.pem', 'wb') as f:
    f.write(private_pem)
# B的公钥对的生成并保存
public_pem = rsa.publickey().exportKey()
with open('bbb-public.pem', 'wb') as f:
    f.write(public_pem)
# 加密和解密
print ("2、加密和解密")
#A发送消息给B，用对方的公钥加密消息
# A使用B的公钥对内容进行rsa 加密
message = '大家好，这就是我要加密的数据'
print ("message: " + message)
#读取文件获得B的公钥
with open('bbb-public.pem') as f:
    key = f.read()
rsakey = RSA.importKey(str(key))
cipher = Cipher_pkcs1_v1_5.new(rsakey)
cipher_text = base64.b64encode(cipher.encrypt(bytes(message.encode("utf8"))))
print ("加密(encrypt)")
print (cipher_text)
# B使用自己的私钥对内容进行rsa 解密
with open('bbb-private.pem') as f:
    key = f.read()
rsakey = RSA.importKey(key)
cipher = Cipher_pkcs1_v1_5.new(rsakey)
text = cipher.decrypt(base64.b64decode(cipher_text), random_generator)
print( "解密(decrypt)")
print ("text:" + str(text,"utf8"))
print("message:"+message)
assert str(text,"utf8") == message
```

```python
#PKI仿真
'''
--setup
1 CA
2 用户注册
3 用户验证签名
--running
1 Alice准备好要发送的消息明文M。
2 Alice对消息M做哈希运算，产生一个消息摘要ℎ_M,A。
3 Alice用自己的私钥SK_A对消息摘要h做数字签名得到σ_A_h_M,A。
4 Alice随机产生一个对称加密密钥K（例如AES密钥），并用此密钥K对要发送的消息进行加密，得到密文。
5 Alice下载到Bob的证书，并验证其证书的合法性。
6 如果上述验证通过，Alice用Bob的公钥PK_A（从证书中获得）对刚才选取的对称加密密钥K进行加密，得到密文消息C。
Alice将密文消息、加密后的对称加密密钥K以及对消息摘要的签名σ_A_h_M,A 发送给Bob。
7 Bob收到消息后，首先用自己的私钥SK_B解密得到对称密钥K。
8 Bob用密钥K解密得到明文消息M。
9 Bob对消息M做个哈希得到h′_M,A。
10 Bob下载到Alice的证书，同样地对Alice的证书有效性进行验证。
11 如果Alice的证书有效，则Bob用证书里面的公钥对签名σ_A_h_M,A进行解密，得到原始哈希h_M,A。
12 Bob比较两个哈希是否一致，如果一致，则说明收到的消息没有被修改过。
'''
from Crypto import Random
from Crypto.Hash import SHA
from Crypto.Cipher import PKCS1_v1_5 as Cipher_pkcs1_v1_5
from Crypto.Signature import PKCS1_v1_5 as Signature_pkcs1_v1_5

from Crypto.PublicKey import RSA

import base64
from binascii import b2a_hex, a2b_hex

from os import urandom
from Crypto.Cipher import AES

#setup
#1 CA建立
print ("1、CA公钥私钥")

random_generator = Random.new().read
# rsa算法生成实例
rsa = RSA.generate(1024, random_generator)
# A的秘钥对的生成
privateS_pem = rsa.exportKey()
with open('ca-private.pem', 'wb') as f:
    f.write(privateS_pem)

publicS_pem = rsa.publickey().exportKey()
with open('ca-public.pem', 'wb') as f:
    f.write(publicS_pem)

print("CA的公钥为{}，私钥为{}".format(publicS_pem,privateS_pem))
```

```python
# 223428010210 陈梓欣

from Crypto import Random
from Crypto.Hash import SHA
from Crypto.Signature import PKCS1_v1_5 as Signature_pkcs1_v1_5
from Crypto.Cipher import PKCS1_v1_5 as Cipher_pkcs1_v1_5
from Crypto.PublicKey import RSA
import base64

random_generator = Random.new().read

# 1. 生成发送方密钥对（签名用）
signer_rsa = RSA.generate(1024, random_generator)
with open('sender-private.pem', 'wb') as f:
    f.write(signer_rsa.exportKey())
with open('sender-public.pem', 'wb') as f:
    f.write(signer_rsa.publickey().exportKey())

# 2. 生成接收方密钥对（加密用）
receiver_rsa = RSA.generate(1024, random_generator)
with open('receiver-private.pem', 'wb') as f:
    f.write(receiver_rsa.exportKey())
with open('receiver-public.pem', 'wb') as f:
    f.write(receiver_rsa.publickey().exportKey())

print("密钥对生成完毕\n")

# 发送方操作
def sender_process(message: str):
    print("发送方消息:", message)

    # 读取发送方私钥，签名消息
    with open('sender-private.pem') as f:
        sender_private_key = RSA.importKey(f.read())
    signer = Signature_pkcs1_v1_5.new(sender_private_key)

    # 计算消息摘要
    digest = SHA.new(message.encode('utf-8'))

    # 生成签名
    signature = signer.sign(digest)
    signature_b64 = base64.b64encode(signature).decode('utf-8')
    print("生成签名(Base64):", signature_b64)

    # 读取接收方公钥，进行消息加密
    with open('receiver-public.pem') as f:
        receiver_public_key = RSA.importKey(f.read())
    cipher = Cipher_pkcs1_v1_5.new(receiver_public_key)

    encrypted_message = cipher.encrypt(message.encode('utf-8'))
    encrypted_message_b64 = base64.b64encode(encrypted_message).decode('utf-8')
    print("加密后的消息(Base64):", encrypted_message_b64)

    return encrypted_message_b64, signature_b64

# 接收方操作
def receiver_process(encrypted_message_b64: str, signature_b64: str):
    # 读取接收方私钥，解密消息
    with open('receiver-private.pem') as f:
        receiver_private_key = RSA.importKey(f.read())
    cipher = Cipher_pkcs1_v1_5.new(receiver_private_key)

    encrypted_message = base64.b64decode(encrypted_message_b64)
    sentinel = Random.new().read(15)  # 解密失败时返回随机字节
    decrypted_message = cipher.decrypt(encrypted_message, sentinel).decode('utf-8')
    print("\n接收方解密消息:", decrypted_message)

    # 读取发送方公钥，验证签名
    with open('sender-public.pem') as f:
        sender_public_key = RSA.importKey(f.read())
    verifier = Signature_pkcs1_v1_5.new(sender_public_key)

    digest = SHA.new(decrypted_message.encode('utf-8'))
    signature = base64.b64decode(signature_b64)

    if verifier.verify(digest, signature):
        print("✅ 签名验证成功，消息可信")
    else:
        print("❌ 签名验证失败，消息可能被篡改")

# 测试流程
if __name__ == "__main__":
    original_message = "RSA签名加密完整流程测试"
    encrypted_msg, signature = sender_process(original_message)
    receiver_process(encrypted_msg, signature)
```
