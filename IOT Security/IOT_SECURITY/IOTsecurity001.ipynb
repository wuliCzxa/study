```python
# 223428010210 陈梓欣

#蒙特卡洛方法求PI
from random import random
from math import sqrt
#DARTS 总点数，hits圆内点数量
DARTS = 2**20
hits = 0

for i in range(1,DARTS):
    #生成随机点
    x, y = random(), random()
    #计算点到圆心的距离
    dist = sqrt(x**2 + y**2)
    if dist <= 1.0:
        #圆内点hits+1
        hits = hits + 1
pi = 4 * (hits/DARTS)
print("Pi is %s" % pi)
```

```python
import math
print("一元二次方程求解")
a,b,c = eval(input("输入系数a,b,c:"))
delta= b*b-4*a*c
if delta >= 0:
    delta2= math.sqrt(delta)
    root1= (-1*b+delta2)/(2*a)
    root2= (-1*b-delta2)/(2*a)
    print ("方程解为：",root1,root2)
else:
    print("方程无实根")
```

```python
# 223428010210 陈梓欣

print("二元一次方程组求解")
# 输入两个方程的系数 a1, b1, c1 和 a2, b2, c2
a1, b1, c1 = eval(input("输入第一个方程的系数 a1, b1, c1: "))
a2, b2, c2 = eval(input("输入第二个方程的系数 a2, b2, c2: "))
# 计算行列式
D = a1 * b2 - a2 * b1
if D != 0:
    Dx = c1 * b2 - c2 * b1
    Dy = a1 * c2 - a2 * c1
    x = Dx / D
    y = Dy / D
    print("方程组的解为：x =", x, ", y =", y)
else:
    # 检查是否是无解还是无穷解
    if a1 * c2 == a2 * c1 and b1 * c2 == b2 * c1:
        print("方程组有无穷多组解")
    else:
        print("方程组无解")
```
