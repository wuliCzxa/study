```python
# 两种温度转换程序
val = input("请输入温度和温度类型符号(如: 32C,40F): ")
if val [-1] in ['C','c']: 
    f = 1.8 * float(val [0:-1]) + 32 
    print("转换后的温度结果为: %.2fF"%f)
elif val [-1] in ['F','f']: 
    c = (float(val [0:-1]) - 32) / 1.8 
    print("转换后的温度结果为: %.2fC"%c) 
else: 
    print("输入错误！")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.image import imread

# 加载图像，确保路径正确
img_arr = imread('img.jpg')  # 或 .jpg、.bmp 等
img_arr = img_arr.astype(np.int32)  # 为避免负数溢出，先转换类型

# 每个像素值减去30
img_arr = img_arr - 30
img_arr = np.clip(img_arr, 0, 255)  # 保证像素值仍在 [0, 255] 范围

# 显示图像
plt.imshow(img_arr.astype(np.uint8))
plt.axis('off')
plt.show()
```

```python
#数组级联
import numpy as np
arr = np.random.randint(0,100,size=(5,6))
print("arr =", arr)
arr2 = np.ones(shape=(5,6))
print("arr2 =", arr2)
#axis =1 水平拼接，列增
print(np.concatenate((arr,arr2),axis=1))
#axis = 0 垂直拼接，行增
print(np.concatenate((arr,arr2),axis=0))
```

```python
#图像垂直翻转拼接
img_arr2 = img_arr[::-1,:,:]
img_out = np.concatenate((img_arr,img_arr2), axis = 0)
plt.imshow(img_out)
```

```python
#图像水平翻转拼接
img_arr3 = img_arr[:,::-1,:]
img_out1 = np.concatenate((img_arr,img_arr3), axis = 1)
plt.imshow(img_out1)
```

## 方程组 1x+2y = 1 ;  3x+4y = 4

```python
#方程组 1x+2y = 1 3x+4y = 4
#求线性方程
a1 = np.array([[1,2],[3,4]])
A = a1
b = np.array([[1],[4]])
x = np.linalg.solve(A,b)
print(x)
```

```python
#验证结果
c = np.dot(A,x)
print (c)
```
