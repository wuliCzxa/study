```python
# 223428010210 陈梓欣
import tushare as ts
import pandas as pd
from pandas import DataFrame, Series
import numpy as np

# 设置Tushare token（需要先注册获取）
ts.set_token('b3a13a06084db2e3de45cee0e213939c1c4f930595a18c70cdb6659c')  # 替换为你在tushare.pro官网注册的API token

# 初始化pro接口
pro = ts.pro_api()

# 示例1：获取股票日线行情
def get_stock_daily(ts_code='000001.SZ', start_date='20230101', end_date='20231231'):
    """
    获取个股日线行情
    :param ts_code: 股票代码 (e.g. 600000.SH)
    :param start_date: 开始日期 (YYYYMMDD)
    :param end_date: 结束日期 (YYYYMMDD)
    :return: DataFrame
    """
    df = pro.daily(ts_code=ts_code, start_date=start_date, end_date=end_date)
    df['trade_date'] = pd.to_datetime(df['trade_date'])
    df.set_index('trade_date', inplace=True)
    df.sort_index(inplace=True)
    return df

# 示例2：获取股票基本信息
def get_stock_basic():
    """获取沪深上市公司基本情况"""
    df = pro.stock_basic(exchange='', list_status='L', fields='ts_code,symbol,name,industry,list_date')
    return df

# 示例3：技术指标计算与分析
def calculate_technical_indicators(df):
    """
    计算技术指标
    :param df: 包含日线数据的DataFrame
    :return: 添加技术指标后的DataFrame
    """
    # 计算简单移动平均线
    df['MA5'] = df['close'].rolling(window=5).mean()
    df['MA20'] = df['close'].rolling(window=20).mean()
    
    # 计算指数移动平均线
    df['EMA12'] = df['close'].ewm(span=12, adjust=False).mean()
    df['EMA26'] = df['close'].ewm(span=26, adjust=False).mean()
    
    # 计算MACD
    df['DIF'] = df['EMA12'] - df['EMA26']
    df['DEA'] = df['DIF'].ewm(span=9, adjust=False).mean()
    df['MACD'] = 2 * (df['DIF'] - df['DEA'])
    
    # 计算RSI
    delta = df['close'].diff()
    gain = (delta.where(delta > 0, 0)).rolling(window=14).mean()
    loss = (-delta.where(delta < 0, 0)).rolling(window=14).mean()
    rs = gain / loss
    df['RSI'] = 100 - (100 / (1 + rs))
    
    return df

# 示例4：获取指数数据
def get_index_daily(ts_code='000001.SH', start_date='20230101', end_date='20231231'):
    """获取指数日线行情"""
    df = pro.index_daily(ts_code=ts_code, start_date=start_date, end_date=end_date)
    df['trade_date'] = pd.to_datetime(df['trade_date'])
    df.set_index('trade_date', inplace=True)
    return df.sort_index()

# 示例5：数据可视化
def visualize_data(df, title='Stock Analysis'):
    """简单可视化股票数据"""
    import matplotlib.pyplot as plt
    plt.figure(figsize=(14, 8))
    
    # 绘制价格和移动平均线
    plt.subplot(2, 1, 1)
    plt.plot(df.index, df['close'], label='Close Price')
    plt.plot(df.index, df['MA5'], label='5-day MA')
    plt.plot(df.index, df['MA20'], label='20-day MA')
    plt.title(title)
    plt.legend()
    
    # 绘制MACD
    plt.subplot(2, 1, 2)
    plt.bar(df.index, df['MACD'], color=np.where(df['MACD'] > 0, 'r', 'g'))
    plt.plot(df.index, df['DIF'], label='DIF')
    plt.plot(df.index, df['DEA'], label='DEA')
    plt.title('MACD Indicator')
    plt.legend()
    
    plt.tight_layout()
    plt.show()

# 使用示例
if __name__ == "__main__":
    # 获取股票基本信息
    stock_basic = get_stock_basic()
    print("股票基本信息:")
    print(stock_basic.head())
    
    # 获取贵州茅台日线数据
    df = get_stock_daily(ts_code='600519.SH', start_date='20230101', end_date='20231231')
    
    # 计算技术指标
    df = calculate_technical_indicators(df)
    
    # 显示处理后的数据
    print("\n贵州茅台2023年行情数据:")
    print(df.tail())
    
    # 可视化分析
    visualize_data(df, title='贵州茅台 (600519.SH) 2023年走势分析')
    
    # 获取上证指数数据
    sh_index = get_index_daily(ts_code='000001.SH', start_date='20230101', end_date='20231231')
    print("\n上证指数2023年收盘价:")
    print(sh_index[['close']].tail())
```
